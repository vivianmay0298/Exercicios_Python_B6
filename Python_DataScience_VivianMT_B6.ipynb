```python
#1. Calcule a integral da função f(x) = x2 sin(x) no intervalo de 0 a π.

def integ(x):
  return x**2 * np.sin(x)

result, erro = quad(integ, 0, np.pi)
print("Resultado da integral: ", result)
```

```python
#2. Resolva a equação diferencial dy/dx = x−y, com y(0) = 1, no intervalo de 0 a 10.

def difer(x, y):
  return x - y

result = solve_ivp(difer, [0, 10],[1])
print(f"Tempo: \n", result.t)
print(f"\nValor de y: \n", result.y[0])
```

```python
#3. Encontre as raízes da função f(x) = x**4 − 3x**3 + 2 utilizando SciPy.

def f(x):
    return x**4 - 3 * x**3 + 2

num_inicial = [0.0, 3.0]
raizes = fsolve(f, num_inicial)

print("Raízes reais encontradas:")
for raiz in raizes:
    print(f"x = {raiz:.5f}")
```

```python
#5. Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).

t = np.linspace(0,1)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)
frequencias = fft(sinal)

print("Transformada de Fourier:\n", frequencias)
```

```python
#7. Use a função de interpolação cúbica de SciPy para interpolar os dados (0,0),(1,2),(2,3),(3,5),(4,4).

x = np.array([0, 1, 2, 3, 4])
y = np.array([0, 2, 3, 5, 4])

interp = interp1d(x, y, kind='cubic')

x_novos = np.random.randint(0, 4, size=5)
y_novos = interp(x_novos)

print(f"Valores interpolados: {y_novos}")
```

```python
#8. Resolva o sistema de equações 3x + 2y = 5 e x + 4y = 6.

a = np.array([[3, 2],[1, 4]])
b = np.array([5, 6])

x, y = linalg.solve(a, b)
print(f"Resultado: x = {x:.2f} e y = {y}")
```

```python
#9. Ajuste uma curva quadrática aos pontos (1,2),(2,4),(3,5),(4,7),(5,8).

def curva(x, a, b, c):
  return a * x**2 + b * x + c

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])

coef ,erro = curve_fit(curva, x, y)
a, b, c = coef

print(f"Coeficientes:\na = {a:.2f}\nb = {b:.2f}\nc = {c:.2f}")
print(f"\nA equação da curva é: y = {a:.2f}x² + {b:.2f}x + {c:.2f}")
```

```python
#10. Calcule a integral dupla da função f(x,y) = xy + x2 no intervalo [0,2] para x e [0,1] para y.

def integral_dupla(y, x):
  return (x * y) + (x * 2)

result, erro = dblquad(integral_dupla, 0,2, 0, 1)
print("Resultado da integral dupla: ", result)
```

```python
#11. Encontre os autovalores e autovetores de uma matriz 4x4 de números inteiros aleatórios.

np.set_printoptions(precision=2, suppress=True)
matriz = np.random.randint(0,10, size=(4,4))

autovalor, autovetor = np.linalg.eig(matriz)
print("Autovalor da matriz 4x4:\n", autovalor)
print("\nAutovetor da matriz 4x4:\n",autovetor)
```

```python
#14. Utilize a função de otimização multivariada para minimizar f(x,y) = (x − 2)**2 + (y − 3)**2.

def funcao(f):
  x, y = f
  return (x - 2)**2 + (y - 3)**2

result = minimize(funcao, [0, 0])

print(f"Mínimo encontrado em x = {result.x[0]:.0f} e y = {result.x[1]:.0f}")
```

```python
#18. Interpole os valores de uma função f(x) = x**2 nos pontos [1,2,3,4,5] utilizando spline cúbica.

x = np.array([1, 2, 3, 4, 5])
y = x**2

spline = CubicSpline(x, y)

print("Valores de y:", y)

for valor in x:
    print(f"x = {valor} | y = {spline(valor)}")
```

```python
#18. Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

#Criando dados
dados = pd.DataFrame({
    'Produto': ['Televisão', 'Celular', 'Mouse', 'Computador', 'Teclado'],
    'Quantidade': [-10, 25, -15, 20, 18]
})

#Definindo cores em uma lista
cores = ['dodgerblue' if x > 0 else 'crimson' for x in dados['Quantidade']]

#Criando e personalizando um gráfico de barras
grafico = plt.bar(dados['Produto'], dados['Quantidade'], color=cores)

#Personalizando
plt.bar_label(grafico, padding=3, fontweight='light')
plt.ylim(-20, 30)

#Descricao do título e eixos
plt.title('Vendas por Produto')
plt.xlabel('Produto')
plt.ylabel('Quantidade')

plt.show()
```

```python
#19. Plote um gráfico de linha com duas séries e adicione uma legenda para diferenciá-las.

# Criando dados
anos = [2020, 2021, 2022, 2023, 2024]
lucroA = [10, 15, 20, 25, 30]
lucroB = [12, 14, 16, 18, 20]

#Plotando primeira série
plt.plot(anos, lucroA, color='dodgerblue', marker='o', label='Empresa A')

#Plotando segunda série
plt.plot(anos, lucroB, color='crimson', marker='s', label='Empresa B')

#Exibindo a legenda
plt.legend()

# 5. Perfumaria (Títulos e Grade)
plt.title('Comparação de Lucro entre a empresa A e B')
plt.xlabel('Anos')
plt.ylabel('Lucro')
plt.grid(True, linestyle='--', alpha=0.7)

plt.show()
```
