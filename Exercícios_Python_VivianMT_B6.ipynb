21. Busca Binária
* O código realiza uma *busca binária* em uma lista ordenada para encontrar a posição de um determinado número. Inicialmente, define os limites de busca como o primeiro e o último índice da lista. Em seguida, utiliza um laço *while* para calcular o índice do elemento central e compará-lo com o número que está sendo buscado. Se o elemento central for igual ao valor procurado, sua posição é exibida e a busca é encerrada. Caso o elemento central seja menor que o valor procurado, a busca continua na metade direita da lista, caso seja maior, continua na metade esquerda.

```python
lista = [1,2,3,4,5,6,7,8,9]
buscar = 6
inicio = 0
fim = len(lista)-1

while inicio <= fim:
  meio = int((inicio + fim) / 2)

  if lista[meio] == buscar:
    print(f"Número {buscar} na posição {meio}.")
    break
  elif lista[meio] < buscar:
    inicio = meio + 1
  else:
    fim = meio - 1
```

26. Calculadora Simples
* O código implementa uma calculadora que permite ao usuário realizar operações de adição, subtração, multiplicação e divisão. Inicialmente, solicita a escolha da operação e o primeiro número. Em seguida, utiliza um laço while para permitir a realização de vários cálculos consecutivos. A estrutura *match* identifica a operação escolhida e atualiza o valor de 'n1' com o resultado. Durante a divisão, verifica se o segundo número é zero e solicita outro valor caso seja. Após cada operação, o programa pergunta se o usuário deseja realizar um novo cálculo. O processo continua até que a opção '5' seja escolhida. Por fim, exibe o resultado total.

```python
op = int(input("Escolha a operação desejada: \n1 - Adição \n2 - Subtração \n3 - Multiplicação \n4 - Divisão \n"))

if op < 1 or op > 4:
  op = int(input("Opção inválida. Digite uma opção das operações disponíveis: "))

n1 = int(input("N1: "))

while op != 5:
  n2 = int(input("N2: "))

  match op:
    case 1:
      n1 += n2
    case 2:
      n1 -= n2
    case 3:
      n1 *= n2
    case 4:
      if n2 == 0:
        n2 = int(input("Não é possível dividir por zero. Digite outro número: "))
      n1 /= n2

  op = int(input(f"Total = {n1}. \n\nDeseja realizar mais cálculos?(Não = 5) \nSe sim, escolha a operação desejada: \n1 - Adição \n2 - Subtração \n3 - Multiplicação \n4 - Divisão \n"))

  if op < 1 or op > 6:
    op = int(input("Opção inválida. Digite uma opção das operações disponíveis: "))

print(f"Total = {n1}")
```

27. Anagramas
* O código faz um import da biblioteca *string* e é definido duas palavras para verificar se são anagramas. Em seguida, cria *dicionários* para armazenar a quantidade de vezes que cada letra aparece em cada palavra. Utiliza laços *for* para percorrer as letras das palavras e compará-las com as letras do alfabeto, contabilizando suas ocorrências nos respectivos dicionários. Por fim, compara os dois dicionários, se forem iguais, significa que as palavras possuem as mesmas letras nas mesmas quantidades e, portanto, são anagramas. Caso contrário, informa que não são anagramas.

```python
import string

p1 = "calor"
p2 = "coral"

alf = string.ascii_letters
contador = {}
contador2 ={}

for i in p1:
  for j in alf:
    if i == j:
      if i not in contador:
        contador[i] = 0
      contador[i] += 1

for i in p2:
  for j in alf:
    if i == j:
      if i not in contador2:
        contador2[i] = 0
      contador2[i] += 1

if contador == contador2:
  print(f"{p1} e {p2} são anagramas.")
else:
  print(f"{p1} e {p2} não são anagramas.")
```

31. Ordenação por Bolha
* Dado uma lista de números desordenados, o código utiliza o algoritmo de ordenação Bubble Sort para organizá-los em ordem crescente. Primeiro, exibe a lista original. Em seguida, utiliza dois laços for para comparar elementos consecutivos da lista. Quando um elemento é maior que o próximo, os valores são trocados de posição. Esse processo é repetido até que todos os elementos estejam ordenados. Por fim, exibe a lista organizada em ordem crescente.

```python
lista = [5,3,4,1,2,8,7,9,6]
print(lista)

for i in range(len(lista)-1):
  for j in range(len(lista)-1):
    if lista[j] > lista[j + 1]:
      lista[j], lista[j + 1] = lista[j + 1], lista[j]

print(lista)
```

34. Contador de Vogais

```python
import string

palavra = input("Digite uma palavra: ")
vogais = ["a","e","i","o","u"]
ver = []
contador = {}


for letra in palavra:
  for v in vogais:
    if letra == v:
      ver.append(letra)

for v in ver:
  if v not in contador:
    contador[v] = 0
  contador[v] += 1

print(contador)
```

```python
f = int(input("Escolha um limite para a sequência de Fibonacci: "))
x = [0,1]

for j in x:
  if x[len(x)-1] < f:
    x.append(x[len(x)-2] + x[len(x)-1])
  if x[len(x)-1] > f:
    x.remove(x[len(x)-1])

print(x)
```

38. Contar Palavras em Texto

Por fim, utiliza *len()* para contar os elementos restantes e exibe a quantidade de palavras do texto.

```python
import string

numero = string.digits

texto = input("Insira seu texto: ")
separador = texto.split()

for i in separador:
  for n in numero:
    if i == n:
      separador.remove(i)

print(f"O texto possui {len(separador)} palavras.")
```

40. Lista de Compras
* O código implementa uma lista de compras utilizando um dicionário para armazenar os itens e suas respectivas quantidades. Inicialmente, apresenta um menu de opções para o usuário adicionar, remover ou listar itens, além de encerrar o programa. Utiliza um laço *while* para manter o sistema em funcionamento e a estrutura *match* para executar a operação escolhida. Ao adicionar um item, solicita seu nome e quantidade e os armazena no dicionário. Ao remover, verifica se a lista possui itens, exibe os itens disponíveis e permite selecionar um item para exclusão. A opção de listar exibe todos os itens e suas quantidades. O programa continua permitindo novas operações até que a opção de encerramento seja escolhida.

```python
listadecompras = {}
op = int(input("Escolha a operação desejada: \n1 - Adicionar item. \n2 - Remover item. \n3 - Listar itens.\n4 - Encerrar\n"))

if op < 1 or op > 4:
  op = int(input("Opção inválida. Digite uma opção disponível: "))

while op != 4:
  match op:
    case 1:
      chave = input("Item: ")
      valor = int(input("Quantidade: "))
      listadecompras[chave] = valor
      if chave in listadecompras:
        print("Salvo com sucesso!")
      else:
        print("Tente novamente mais tarde!")

    case 2:
      if not listadecompras:
        print("A Lista de Compras está vazia!")

      else:
        print("\nLista de compras")
        for chave, valor in listadecompras.items():
          print(f"{chave}: {valor}")

        excluir = input("\nQual item deseja remover da lista?: ")
        if excluir in listadecompras: del listadecompras[excluir]

    case 3:
      if not listadecompras:
        print("A Lista de Compras está vazia!")
      else:
        print("\nLista de compras")
        for chave, valor in listadecompras.items():
          print(f"{chave}: {valor}")

    case 4:
      break

  op = int(input(f"\nEscolha a operação desejada: \n1 - Adicionar item. \n2 - Remover item. \n3 - Listar itens.\n4 - Encerrar\n"))

  if op < 1 or op > 4:
    op = int(input("\nOpção inválida. Digite uma opção disponíveis: "))

print("Encerrando o sistema!")
```

42. Números Primos até N
* O código solicita um valor numérico ao usuário. Para cada número, utiliza um laço *for* para verificar seus possíveis divisores por meio do operador %. Os divisores encontrados são armazenados na lista 'x'. Caso o número possua exatamente dois divisores, 1 e ele mesmo, ele é considerado primo e adicionado à lista 'listadeprimos'. Após verificar cada número, a lista 'x' é limpa para que o próximo número seja analisado. Por fim, o programa exibe a lista de números primos encontrados até o limite informado.

```python
numerosverificar = int(input("Verificar quais números são primos até: "))
x = []
listadeprimos = []

for numerosprimos in range(1, numerosverificar+1):
  for i in range(1,numerosprimos+1):

    ver = numerosprimos%i
    if ver == 0:
      x.append(i)

  if len(x) == 2:
    listadeprimos.append(numerosprimos)

  x.clear()

print(f"De 1 a {numerosverificar} esta é a lista de números primos = {listadeprimos}")
```

44. Fibonacci em Nível
* O código solicita ao usuário a quantidade de termos desejada da sequência de Fibonacci e inicia uma lista com os valores 0 e 1. Em seguida, utiliza um laço *while* para verificar se a quantidade de elementos da lista corresponde à quantidade de termos solicitada. Caso seja solicitado apenas um termo, remove o valor 1, mantendo somente o primeiro termo da sequência. Para quantidades maiores, utiliza um laço *for* para calcular os próximos termos somando os dois últimos valores da lista e adicionando o resultado ao final. Por fim, exibe a sequência de Fibonacci com a quantidade de termos solicitada.

```python
t = int(input("Digite a quantidade de termos de Fibonacci desejada: "))
x = [0,1]

while len(x) != t:
  if t == 1:
    x.remove(1)
  else:
    for i in range(0, t-2):
      x.append(x[len(x)-2] + x[len(x)-1])

print(x)
```

45. Par ou Ímpar em Lista
* Dada uma lista contendo números inteiros e duas listas vazias para separar os números pares e ímpares. 47. Contar Números em Lista
* Dada uma lista de números e um dicionário vazio chamado 'contador' para armazenar a frequência de cada valor. O código utiliza um laço "for" para percorrer cada número da lista. Para cada número, verifica se ele ainda não está no dicionário, caso não esteja, cria uma chave para ele com valor inicial 0. Depois, incrementa o valor em 1, contabilizando quantas vezes cada número aparece. Por fim, exibe o dicionário com a frequência de cada número da lista.

```python
lista = [1,2,3,3,5,2,6,6,8,9,9,9]
contador = {}

for num in lista:
  if num not in contador:
    contador[num] = 0
  contador[num] += 1

print(contador)
```

48. Ordenação por Inserção
* Dada uma determinada lista de números desordenados, o código utiliza o algoritmo de ordenação Insertion Sort para organizá-los em ordem crescente. Primeiro, exibe a lista original. Em seguida, utiliza um laço *for* para percorrer os elementos da lista a partir do segundo elemento, armazenando o valor atual e comparando-o com os elementos anteriores. Enquanto o elemento anterior for maior que o valor atual, os elementos são deslocados uma posição para a direita. O valor então é inserido na posição correta. Por fim, o programa exibe a lista ordenada em ordem crescente.

```python
lista = [5,3,4,1,2,8,7,9,6]
print(lista)

for i in range(1,len(lista)):
  valor = lista[i]
  j = i - 1

  while j >= 0 and lista[j] > valor:
    lista[j + 1] = lista[j]
    j -= 1
    lista[j + 1] = valor

print(lista)
```

50. Cálculo de Frequência de Letras
* O código fazo import da biblioteca *string* e solicita ao usuário uma palavra. Em seguida, utiliza *string.ascii_letters* para obter as letras do alfabeto e cria um dicionário vazio chamado 'contador' para armazenar a frequência de cada letra. Utiliza dois laços *for* para percorrer as letras da palavra e compará-las com as letras do alfabeto. Quando encontra uma correspondência, verifica se a letra já está no dicionário e, caso não esteja, cria uma chave com valor inicial 0. Depois, incrementa o contador em 1. Por fim, exibe o dicionário contendo a quantidade de vezes que cada letra aparece na palavra.

```python
import string

palavra = input("Digite uma palavra: ")
alf = string.ascii_letters
contador = {}

for i in palavra:
  for j in alf:
    if i == j:
      if i not in contador:
        contador[i] = 0
      contador[i] += 1

print(contador)
```
